```python
import pandas as pd

COCONUT_ZIP = "data/coconut/coconut_csv-10-2026.zip"

# 44 columns in the full release; these are the ones most useful for screening.
# Drop usecols to pull all of them (names, synonyms, dois, cas, full taxonomy).
cols = [
    "identifier", "canonical_smiles", "name",
    "molecular_weight", "alogp", "topological_polar_surface_area",
    "hydrogen_bond_acceptors", "hydrogen_bond_donors",
    "lipinski_rule_of_five_violations",
    "qed_drug_likeliness", "np_likeness",
    "np_classifier_pathway", "np_classifier_superclass", "np_classifier_class",
    "organisms",
]

coconut = pd.read_csv(COCONUT_ZIP, usecols=cols, low_memory=False)

print(coconut.shape)
print(coconut.iloc[:10])
```
